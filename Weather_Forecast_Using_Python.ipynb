```python
import requests

weather_descriptions = {
    0: "Clear sky",
    1: "Mainly clear",
    2: "Partly cloudy",
    3: "Overcast",
    45: "Fog",
    48: "Depositing rime fog",
    51: "Light drizzle",
    53: "Moderate drizzle",
    55: "Dense drizzle",
    61: "Slight rain",
    63: "Moderate rain",
    65: "Heavy rain",
    71: "Slight snow fall",
    73: "Moderate snow fall",
    75: "Heavy snow fall",
    80: "Rain showers",
    95: "Thunderstorm",
}
```

```python
def get_coordinates(city):
    try:
        lat_long = f"https://geocoding-api.open-meteo.com/v1/search?name={city}&count=1"
        response = requests.get(lat_long)
        response.raise_for_status()
        data = response.json()
        if "results" not in data or not data["results"]:
            print("City not found!")
            return None
        location = data["results"][0]
        return location.get("latitude"), location.get("longitude"), location.get("name"), location.get("country")
    except requests.exceptions.RequestException as e:
        print(f"Request error: {e}")
        return None
```

```python
def get_weather_forecast(lat, lon):
    try:
        weather_url = (
            f"https://api.open-meteo.com/v1/forecast?latitude={lat}&longitude={lon}&daily=temperature_2m_max,temperature_2m_min,weathercode&timezone=Asia/Kolkata"
        )
        response = requests.get(weather_url)
        response.raise_for_status()
        data = response.json()

        if "daily" not in data:
            print("No daily forecast data available.")
            return

        daily_data = data.get("daily", {})
        dates = daily_data.get("time", [])
        max_temps = daily_data.get("temperature_2m_max", [])
        min_temps = daily_data.get("temperature_2m_min", [])
        weather_codes = daily_data.get("weathercode", [])

        if dates and max_temps and min_temps and weather_codes:
            print("Weather Forecast:\n")
            for i in range(min(2, len(dates))):
                code = weather_codes[i]
                description = weather_descriptions.get(code, "Unknown weather")
                print(f"Date: {dates[i]}")
                print(f"Max Temp: {max_temps[i]}°C")
                print(f"Min Temp: {min_temps[i]}°C")
                print(f"Condition: {description}\n")
        else:
            print("No daily forecast data available.")

    except requests.exceptions.RequestException as e:
        print(f"Request error: {e}")
```

```python
import requests
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from datetime import datetime

def get_weather_data(lat, lon):
    weather_url = (
        f"https://api.open-meteo.com/v1/forecast?latitude={lat}&longitude={lon}&daily=temperature_2m_max,temperature_2m_min&timezone=Asia/Kolkata"
    )
    response = requests.get(weather_url)
    return response.json()

def plot_temperature_forecast(lat, lon):
    data = get_weather_data(lat, lon)
    daily_data = data.get("daily", {})
    dates = [datetime.strptime(date, '%Y-%m-%d') for date in daily_data.get("time", [])]
    max_temps = daily_data.get("temperature_2m_max", [])
    min_temps = daily_data.get("temperature_2m_min", [])

    if dates and max_temps and min_temps:
        plt.figure(figsize=(10, 6))
        plt.plot(dates, max_temps, marker='o', label='Max Temp (°C)', color='tomato')
        plt.plot(dates, min_temps, marker='o', label='Min Temp (°C)', color='skyblue')
        plt.title('Temperature Forecast')
        plt.xlabel('Date')
        plt.ylabel('Temperature (°C)')
        plt.legend()
        plt.grid(True)
        plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%Y-%m-%d'))
        plt.gcf().autofmt_xdate()
        plt.show()
    else:
        print("No temperature forecast data available.")
```

```python
import matplotlib.pyplot as plt
from collections import Counter

def plot_weather_condition_pie(descriptions):
    """
    Plots a pie chart showing the distribution of weather conditions.

    Args:
        descriptions (list): List of weather condition descriptions.
    """

    condition_counts = Counter(descriptions)


    labels = list(condition_counts.keys())
    sizes = list(condition_counts.values())

    plt.figure(figsize=(6, 6))
    plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=90, colors=plt.cm.Paired.colors)
    plt.title("Weather Condition Distribution")
    plt.tight_layout()
    plt.show()
```

```python
weather_descriptions = ["Sunny", "Cloudy", "Rainy", "Sunny", "Cloudy", "Sunny"]
plot_weather_condition_pie(weather_descriptions)
```

```python
import requests
import statistics
import numpy as np

weather_descriptions = {
    0: "Clear sky",
    1: "Mainly clear",
    2: "Partly cloudy",
    3: "Overcast",
    45: "Fog",
    48: "Depositing rime fog",
    51: "Light drizzle",
    53: "Moderate drizzle",
    55: "Dense drizzle",
    61: "Slight rain",
    63: "Moderate rain",
    65: "Heavy rain",
    71: "Slight snow fall",
    73: "Moderate snow fall",
    75: "Heavy snow fall",
    80: "Rain showers",
    95: "Thunderstorm",
}

def get_coordinates(city):
    lat_long = f"https://geocoding-api.open-meteo.com/v1/search?name={city}&count=1"
    response = requests.get(lat_long)
    data = response.json()
    if "results" not in data or not data["results"]:
        print("City not found!")
        return None
    location = data["results"][0]
    return location["latitude"], location["longitude"], location["name"], location["country"]

def get_weather_forecast(lat, lon):
    weather_url = (
        f"https://api.open-meteo.com/v1/forecast?latitude={lat}&longitude={lon}&daily=temperature_2m_max,temperature_2m_min,weathercode&timezone=Asia/Kolkata"
    )
    response = requests.get(weather_url)
    data = response.json()
    if "daily" not in data:
        print("No daily forecast data available.")
        return None
    dates = data["daily"]["time"]
    max_temps = data["daily"]["temperature_2m_max"]
    min_temps = data["daily"]["temperature_2m_min"]
    weather_codes = data["daily"]["weathercode"]
    print("Weather Forecast:\n")
    for i in range(min(2, len(dates))):  # Today and Tomorrow
        code = weather_codes[i]
        description = weather_descriptions.get(code, "Unknown weather")
        print(f" Date: {dates[i]}")
        print(f" Max Temp: {max_temps[i]}°C")
        print(f" Min Temp: {min_temps[i]}°C")
        print(f" Condition: {description}\n")
    return max_temps, min_temps, weather_codes

def calculate_temperature_stats(max_temps, min_temps):
    max_temp_mean = statistics.mean(max_temps)
    max_temp_median = statistics.median(max_temps)
    min_temp_mean = statistics.mean(min_temps)
    min_temp_median = statistics.median(min_temps)

    print("Temperature Statistics:")
    print(f"Max Temp Mean: {max_temp_mean}°C")
    print(f"Max Temp Median: {max_temp_median}°C")
    print(f"Min Temp Mean: {min_temp_mean}°C")
    print(f"Min Temp Median: {min_temp_median}°C")

def calculate_weather_condition_frequency(weather_codes):
    weather_freq = {}
    for code in weather_codes:
        description = weather_descriptions.get(code, "Unknown weather")
        if description in weather_freq:
            weather_freq[description] += 1
        else:
            weather_freq[description] = 1

    print("Weather Condition Frequency:")
    for condition, freq in weather_freq.items():
        print(f"{condition}: {freq}")

def calculate_temperature_correlation(max_temps, min_temps):
    correlation = np.corrcoef(max_temps, min_temps)[0, 1]
    print(f"Correlation between Max and Min Temperatures: {correlation}")
```

```python
def main():
    city = input("Enter city name: ")
    result = get_coordinates(city)
    if result:
        lat, lon, city_name, country = result
        print(f"\n Location: {city_name}, {country}")
        forecast_data = get_weather_forecast(lat, lon)
        if forecast_data:
            max_temps, min_temps, weather_codes = forecast_data
            calculate_temperature_stats(max_temps, min_temps)
            calculate_weather_condition_frequency(weather_codes)
            calculate_temperature_correlation(max_temps, min_temps)

if __name__ == "__main__":
    main()
```
